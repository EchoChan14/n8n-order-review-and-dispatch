```python
import os
import pandas as pd
import requests
import datetime

# =====================================================================
# 1. 環境參數設定 (已對齊您的新 Test URL)
# =====================================================================
WEBHOOK_URL = "http://localhost:5678/webhook-test/n8ntest" 

FOLDER_PATH = r"C:\Users\echo6\Downloads\模擬公司-20250829T033655Z-1-001\中小企業的數位化與自動化流程模擬資料"
EXCEL_FILE_NAME = "SME_Full_Simulation_Dataset_v2.xlsx"
SHEET_NAME = "訂單資料"

EXCEL_FILE_PATH = os.path.join(FOLDER_PATH, EXCEL_FILE_NAME)

# =====================================================================
# 2. 載入資料並擷取第 1 筆
# =====================================================================
try:
    df = pd.read_excel(EXCEL_FILE_PATH, sheet_name=SHEET_NAME, engine="openpyxl")
    # 抓取第一筆，並將空值轉換為 None (JSON 的 null)
    raw_first_order = df.head(1).where(pd.notnull(df.head(1)), None).to_dict('records')[0]
    print(f"📊 [成功] 已成功連結本地資料庫！")
except Exception as e:
    print(f"❌ [錯誤] 無法讀取 Excel 檔案: {e}")
    raise

# =====================================================================
# 3. 核心優化：深度清洗資料型態 (將時間物件與 numpy 型態轉為 JSON 安全格式)
# =====================================================================
order_payload = {}
for key, value in raw_first_order.items():
    # 狀況 A：如果值是 Pandas 的 Timestamp 或是 Python 的 datetime 時間物件
    if isinstance(value, (pd.Timestamp, datetime.datetime, datetime.date)):
        order_payload[key] = str(value)  # 強制轉為標準字串，如 "2026-05-16 00:00:00"
    
    # 狀況 B：如果值是 numpy 的特殊數字型態 (如 int64, float64)
    elif hasattr(value, 'item'):
        order_payload[key] = value.item() # 轉為 Python 原生 int/float
        
    # 狀況 C：其餘標準資料型態直接保留
    else:
        order_payload[key] = value

# 動態辨識訂單編號（自動抓取字典中前兩個欄位來顯示，避免寫死欄位名抓到 None）
sample_keys = list(order_payload.keys())[:2]
log_label = ", ".join([f"{k}: {order_payload[k]}" for k in sample_keys])
print(f"🚀 準備發送第 1 筆測試訂單數據 ({log_label})")

# =====================================================================
# 4. 發送乾淨的數據至 n8n
# =====================================================================
print("\n⚡ 正在發送單筆數據至 n8n 測試通道...")
print("="*50)

try:
    # 發送 POST 請求
    response = requests.post(WEBHOOK_URL, json=order_payload, timeout=10)
    
    print(f"✅ [發送成功] n8n 接收端已完美回應！")
    print(f"🔹 HTTP 狀態碼: {response.status_code}")
    print(f"📥 n8n 最終回傳內容: {response.text}")

except requests.exceptions.Timeout:
    print(f"❌ [逾時] n8n 超過 10 秒未回傳。")
    print(f"💡 專家提示：這代表 n8n 已收到資料，但流程可能卡在後續節點（例如等待手動審核），導致沒走到 'Respond to Webhook'。")
except Exception as e:
    print(f"❌ [連線異常]: {e}")

print("="*50)
```
